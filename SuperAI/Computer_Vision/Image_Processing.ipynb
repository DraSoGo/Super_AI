# Poisson blending
- Image = 2d function I(x,y)
- We want to seamlessly paste an object into another image
- Instead of copy the raw pixels value, we consider the change in pixels value, i.e. gradient information

```python
x = np.arange(20)
y = 2 + np.random.rand(20)

z = np.zeros(20)
z[5:10] = np.arange(5,10) + np.random.rand(5)
z[10:15] = 19 - np.arange(10, 15) + np.random.rand(5)

fig, axs = plt.subplots(ncols=2, figsize=(10,4))

axs[0].set_ylim([0,10])
axs[0].bar(x, y)

axs[1].bar(x,z, color='r')
```

```python
a = y.copy()
a[5:15] = z[5:15]

b = y.copy()
b[5:15] = z[5:15] - z[5] + y[5]

c = y.copy()
c[5:15] = z[5:15] - z[14] + y[14]

d = y.copy()
d[5:15] = z[5:15] + 0.5*(y[5]- z[5])  + 0.5*(y[14] - z[14])

fig, axs = plt.subplots(ncols=2, nrows=2, figsize=(10,4))

axs[0,0].set_ylim([0,10])
axs[0,0].bar(x, a)

axs[0,1].set_ylim([0,10])
axs[0,1].bar(x,b, color='r')

axs[1,0].set_ylim([0,10])
axs[1,0].bar(x,c, color='g')

axs[1,1].set_ylim([0,10])
axs[1,1].bar(x,d, color='b')
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

def generate_gabor_data(size=101, sigma=20.0, theta=np.pi/6, lambd=8.0, gamma=1.0, psi=0):
    """Generates X, Y, and Z data for a Gabor kernel."""
    w = size // 2
    h = size // 2
    y, x = np.meshgrid(np.arange(-h, h + 1), np.arange(-w, w + 1))

    # Rotation transformation
    x_theta = x * np.cos(theta) + y * np.sin(theta)
    y_theta = -x * np.sin(theta) + y * np.cos(theta)

    # Gabor calculation
    gaussian_env = np.exp(-0.5 * (x_theta**2 + (gamma**2 * y_theta**2)) / (sigma**2))
    sinusoid = np.cos(2 * np.pi * x_theta / lambd + psi)
    z = gaussian_env * sinusoid

    # Generate the 2D kernel
    w = size // 2
    y, x = np.meshgrid(np.arange(-w, w + 1), np.arange(-w, w + 1))
    x_theta = x * np.cos(theta) + y * np.sin(theta)
    y_theta = -x * np.sin(theta) + y * np.cos(theta)

    gb = np.exp(-.5 * (x_theta**2 + (gamma**2 * y_theta**2)) / (sigma**2)) * np.cos(2 * np.pi * x_theta / lambd)

    return x, y, z, gb

# 1. Generate the data
# We use a small 'lambd' (8.0) compared to 'sigma' (20.0) to get multiple ridges
X, Y, Z, gb = generate_gabor_data(size=11, sigma=5, lambd=4)

# 2. Create the 3D Plot
fig = plt.figure(figsize=(12, 8))
ax = fig.add_subplot(111, projection='3d')

# Plot the surface
surf = ax.plot_surface(X, Y, Z, cmap='viridis', edgecolor='none', alpha=0.9)

# 3. Aesthetics
ax.set_title('3D Gabor Kernel (Multiple Ridges and Valleys)', fontsize=15)
ax.set_xlabel('X axis')
ax.set_ylabel('Y axis')
ax.set_zlabel('Amplitude')
fig.colorbar(surf, ax=ax, shrink=0.5, aspect=10)

# Adjust the viewing angle to see the ridges clearly
ax.view_init(elev=35, azim=-45)

plt.show()
```

```python
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
```

```python
class CNN(nn.Module):
    def __init__(self):
        super(CNN, self).__init__()
        self.conv11 = nn.Conv2d(1, 8, kernel_size=3, padding=1)  # input 1 channel, 8 filters = output 8 channels, output same size as input
        self.bn11   = nn.BatchNorm2d(8)
        self.conv12 = nn.Conv2d(8, 8, kernel_size=3, padding=1)
        self.bn12   = nn.BatchNorm2d(8)
        self.conv13 = nn.Conv2d(8, 8, kernel_size=3, padding=1)
        self.bn13   = nn.BatchNorm2d(8)

        self.conv21 = nn.Conv2d(8, 10, kernel_size=3, padding=1)
        self.conv22 = nn.Conv2d(10, 10, kernel_size=3, padding=1)
        self.conv23 = nn.Conv2d(10, 10, kernel_size=3, padding=1)

        self.fc1 = nn.Linear(490, 50)
        self.fc2 = nn.Linear(50, 10)

    def forward(self, x0): # x0 -> (batch_size, 1, 28, 28)
        x1 = F.relu(self.bn11(self.conv11(x0))) # x1 -> (batch_size, 8, 28, 28)

        x = F.relu(self.bn12(self.conv12(x1)))
        x = F.relu(self.bn13(self.conv13(x))) # x -> (batch_size, 8, 28, 28)

        x = x + x1 # x -> (batch_size, 8, 28, 28)

        x = F.max_pool2d(x, 2) # x -> (batch_size, 8, 14, 14)

        # size 14x14
        x = F.relu(self.conv21(x)) # x -> (batch_size, 10, 14, 14)
        x = F.relu(self.conv22(x))
        x = F.relu(self.conv23(x))
        x = F.max_pool2d(x, 2) # x -> (batch_size, 10, 7, 7)

        # "flatten" >> arrange all values on the feature maps as a single vector >> 490 values
        x = x.view(-1, x.size(dim=1)*x.size(dim=2)*x.size(dim=3))
        # x = x.view(-1, 490)

        x = F.relu(self.fc1(x))
        # x = F.dropout(x, training=self.training) # dropout apply only during training
        x = self.fc2(x)
        return x # F.log_softmax(x, dim=1)
```

```python
model = CNN()

x = torch.rand((8, 1, 28, 28), requires_grad=True)
y = model(x)
y
```

```python
# training config
n_epochs = 3
learning_rate = 0.001
momentum = 0.5
log_interval = 100
```

```python
#from tqdm import tqdm
from tqdm import tqdm_notebook as tqdm

def train_epoch(model, optimizer, data_loader, device):

  loss_fct = nn.CrossEntropyLoss()

  model.train()
  model.to(device)
  train_loss = 0
  for batch_idx, (data, target) in enumerate(pbar := tqdm(data_loader)):
    data, target = data.to(device), target.to(device)

    optimizer.zero_grad()

    output = model(data)              # predict output for current data
    loss = loss_fct(output, target)
    loss.backward()                   # backprop to get gradient

    optimizer.step()                  # update the model

    train_loss += loss

    if batch_idx % log_interval == 0:
      pbar.set_description('Train Epoch: {} [{}/{} ({:.0f}%)]\tLoss: {:.6f}'.format(
        epoch, batch_idx * len(data), len(data_loader.dataset),
        100. * batch_idx / len(data_loader), loss.item()))

  return train_loss
```

```python
def test(model, data_loader, device):
  model.eval()
  model.to(device)
  test_loss = 0
  correct = 0
  with torch.no_grad(): # no need for gradient >>not allocate additional variables
    for data, target in data_loader:
      data, target = data.to(device), target.to(device)
      output = model(data)
      test_loss += F.nll_loss(output, target, size_average=False).item()
      pred = output.data.max(1, keepdim=True)[1]
      correct += pred.eq(target.data.view_as(pred)).sum()
  test_loss /= len(data_loader.dataset)
  test_acc = 100. * correct / len(data_loader.dataset)
  return test_acc.item(), test_loss
```

```python
device = "cuda" if torch.cuda.is_available() else "cpu"
device
```

```python
from tqdm import trange

network = CNN()
optimizer = optim.Adam(network.parameters(), lr=1e-3)

best_acc = 0
for epoch in (pbar := trange(1, n_epochs + 1)):
  print(best_acc, flush=True)

  train_epoch(network, optimizer, train_loader, device)
  test_acc, test_loss = test(network, test_loader, device)

  if test_acc > best_acc:
    pbar.set_description('Test accracy improves from {:.2f} to {:.2f}\n'.format(best_acc, test_acc))
    best_acc = test_acc

    torch.save(network.state_dict(), 'model.pth')
    torch.save(optimizer.state_dict(), 'optimizer.pth')
```

```python
import cv2
import numpy as np

def deskew(img):
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    # 1. Threshold/Edge detection
    # We invert the image if it's white text on black, but usually,
    # Canny works well for finding the boundaries of text lines.
    edges = cv2.Canny(gray, 50, 150, apertureSize=3)

    # 2. Hough Line Transform
    # rho=1, theta=np.pi/180 (1 degree resolution), threshold=100
    lines = cv2.HoughLinesP(edges, 1, np.pi / 180, 100, minLineLength=100, maxLineGap=10)

    angles = []
    for line in lines:
        x1, y1, x2, y2 = line[0]
        # Calculate angle in radians, then convert to degrees
        angle = np.rad2deg(np.arctan2(y2 - y1, x2 - x1))
        angles.append(angle)

    # 3. Filter and find the dominant angle
    # We ignore extreme angles to avoid vertical lines or noise
    dominant_angle = np.median(angles)

    # 4. Rotate the image
    (h, w) = img.shape[:2]
    center = (w // 2, h // 2)
    M = cv2.getRotationMatrix2D(center, dominant_angle, 1.0)
    rotated = cv2.warpAffine(img, M, (w, h), flags=cv2.INTER_CUBIC, borderMode=cv2.BORDER_REPLICATE)

    return rotated, dominant_angle
```
